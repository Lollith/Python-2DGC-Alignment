```python
from interface_jupyter import Interface
import ipywidgets as widgets
import os
```

```python
class H5_cleaner(Interface):
    def __init__(self):
        super().__init__(supported_extensions=('.h5',))
        self._create_h5_cleaner_widget()

        self._create_base_widgets()
        # self._create_action_widgets() 

        self._setup_callbacks()
        self._setup_environment()


    def _create_base_widgets(self):
        """Create base widgets for file selection."""
        self._create_path_widgets()
         
    def _create_h5_cleaner_widget(self):
        self.txt_title = widgets.HTML(value="<H1>Files cleaner</H1>")
        self.run_button, self.clear_button, self.output = self._create_action_widgets()

    def delete_h5_files(self, file, display_path):
        try:
            os.remove(file)
            print(f"Supprime: /{display_path}")
        except Exception as e:
            print(f"erreur lors de la suppression de {display_path} : {e}")


    def _on_button_click(self, b):
        """Handle button click event."""
        
        with self.output:
            self.output.clear_output()
            print("Button clicked")
            print("\n Collecting files from selections...   ")
            selected_files = self.get_all_files_from_selections()
            if not selected_files:
                print("Please select files or folders containing .h5 files.")
                return
            #display
            print(f"\n✅ {len(selected_files)} compatible files found")
            for i, f in enumerate(selected_files, 1):
                if f.startswith(self.docker_volume_path):
                    display_path = f.replace(self.docker_volume_path, "")
                else:
                    display_path = f
                print(f"  {i}. /{display_path}")

                self.delete_h5_files(f, display_path)
            print("H5 deleted")

    def display(self):
        """Display the interface."""
        display(self.txt_title,
                widgets.VBox([self._vbox, self._vbox2]),
                widgets.HBox([self.run_button, self.clear_button]),
                self.output)


```

```python
c = H5_cleaner()
c.display()
```
